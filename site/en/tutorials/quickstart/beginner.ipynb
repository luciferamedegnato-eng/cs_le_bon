```
def main():
  while True:

    try:

       n = int(input("entrez un nombre entier n :"))
       break

    except ValueError:
      print("entrée invalide. Veuillez saisir un nombre entier.")
  somme = 0
  nb_paire = 0
  nb_impaie = 0
  for i in range(1, n+1):
    if i % 2 == 0:
      print(f"{i} est paire")
      nb_paire += 1
    else:
      print(f"{i} est impaire")
      nb_paire += 1
    somme += i
  print(f"\nsomme des nombres de 1 à {n} : {somme}")
  print(f"nombre de nombres paire de 1 à {n} : {nb_paire}")
  print(f"nombre de nombres impaire de 1 à {n} : {nb_impaie}")

if __name__ == "__main__":
  main()
```
